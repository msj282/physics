```python
import math
```

```python
# 초기속도 입력
v0= float(input())
print("초기속도 : ", v0, "m/s")
```

```python
#초기 각도 입력
angle_degree = float(input())
print("초기 각도 : ",angle_degree, "도")

#각도를 radian으로 바꾸기
angle_radian = math.radians(angle_degree)
print("초기 각도 : ", angle_radian, "radian")
```

```python
# x 성분의 속도를 구하시요
v0x = math.cos(angle_radian) * v0
print("x축의 초기속도 : ", v0x , "m/s")
```

```python
# y 성분의 속도를 구하시요
v0y = math.cos(angle_radian) * v0
print("y축의 초기속도 : ", v0y , "m/s")
```

```python
#최고점의 높이를 구하세요
g = 9.8
h = (v0y ** 2 ) / (2 * g)
print("최고점 높이:", h, "m")
```

```python
#최고점에 도착하는 시간을 구하세요
t_top = v0y/g
print("최고점에 도착하는 시간:",t_top, "초")
```

```python
#수평으로 이동한 최고 거리를 구하세요
t_total = 2 * t_top
range_ = v0x * t_total
print("수평으로 이동한 최고 거리", range_, "m")
```

```python
#지면에 닿을 때의 vx, vy, v를 구하세요
vx = v0x
vy = -v0y
vf = math.sqrt(vx ** 2 + vy ** 2)
print("지면에 닿을 때의 vx :", vx, "m/s")
print("지면에 닿을 때의 vy :", vy, "m/s")
print("지면에 닿을 때의 vf :", vf, "m/s")
```
